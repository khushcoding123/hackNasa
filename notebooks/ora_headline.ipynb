# ORA headline figure

Terrain-limited networks break differently from idealised ones. This notebook redraws the numbers written by `python -m ora demo` (`docs/figures/ora_headline.json`). Run `make demo` first.

```python
import json
from pathlib import Path

data = json.loads(Path('../docs/figures/ora_headline.json').read_text())
print(data['scenario'], data['sites'], 'candidate sites,', data['seeds'], 'Monte Carlo seeds')
for name, row in data['graphs'].items():
    print(f"{name:30s} links {row['links']:6d}  articulation points {row['articulation_points']:3d}  "
          f"Schneider R random {row['schneider_r_random']:.3f}  targeted {row['schneider_r_betweenness']:.3f}")
```
